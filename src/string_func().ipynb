```python
%sql
-- Create a sample table
CREATE OR REPLACE TABLE employee (
    emp_id INT,
    emp_name STRING,
    email STRING,
    department STRING
);

-- Insert sample data
INSERT INTO employee
VALUES
(1, 'Roshini Narayanan', 'ROSHINI@GMAIL.COM', '  Data Engineering  '),
(2, 'Rahul Sharma', 'RAHUL@GMAIL.COM', '  Software Development  '),
(3, 'Priya Kumar', 'PRIYA@GMAIL.COM', '  Business Analytics  ');
```

```python
%sql
select upper(emp_name) as employeeName from employee
```

```python
%sql
SELECT
    email,
    LOWER(email) AS lower_email
FROM employee;
```

```python
%sql
SELECT
    emp_name,
    LENGTH(emp_name) AS total_characters
FROM employee;
```

```python
%sql
SELECT
    CONCAT_WS( ' - ', emp_name,department,email) AS employee_details
FROM employee;
```

```python
%sql
SELECT
    emp_name,
    SUBSTRING(emp_name, 1, 7) AS first_part
FROM employee;
```

```python
%sql
SELECT
    department,
    TRIM(department) AS trimmed_department
FROM employee;
```

```python
%sql
SELECT
    department,
    REPLACE(department, 'Engineering', 'Engineer') AS updated_department
FROM employee;
```

```python
%sql
SELECT
    emp_name,
    SPLIT(emp_name, ' ') AS name_parts
FROM employee;
```
