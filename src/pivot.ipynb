```python
%sql
CREATE OR REPLACE TABLE pivot_practice
(
    emp_id INT,
    employee STRING,
    department STRING,
    year INT,
    salary INT
);
```

```python
%sql
INSERT INTO pivot_practice VALUES
(1, 'Ravi', 'HR', 2024, 50000),
(1, 'Ravi', 'Manager', 2025, 60000),
(2, 'Priya', 'HR', 2024, 55000),
(2, 'Priya', 'HR', 2025, 65000),
(3, 'John', 'IT', 2024, 70000),
(3, 'John', 'IT', 2025, 75000);
```

```python
%sql
select * from pivot_practice
```

```python
%sql
select * from pivot_practice
pivot
(
    sum(salary)
    for year in (2024,2025)
)
```
